```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/raw/AllCombined_MentalEffortFocus.csv")
```

```python
features = ['ACCx','ACCy','ACCz','Movement','EDAz','TEMPz','HRz']

for f in features:
    plt.figure(figsize=(12,4))
    plt.plot(df[f])
    plt.title(f)
    plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Age'], bins=15)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")

plt.show()
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns

for col in categorical_cols:
    plt.figure(figsize=(8,5))

    df[col].value_counts().plot(kind='bar')

    plt.xlabel(col)
    plt.ylabel("Count")
    plt.title(f"{col} Distribution")

    plt.xticks(rotation=45)
    plt.show()
```

```python
df.columns
```

```python
cols = ['Puzzle','Focus01','FocusLevel','DifficultyLevel','Correct']

for col in cols:
    plt.figure(figsize=(8,5))

    df[col].value_counts().sort_index().plot(kind='bar')

    plt.xlabel(col)
    plt.ylabel("Count")
    plt.title(f"{col} Distribution")
    plt.xticks(rotation=45)

    plt.show()
```
